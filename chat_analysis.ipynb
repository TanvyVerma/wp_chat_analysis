```python
import re
import pandas as pd 
```

```python
# f = open('WhatsApp Chat with Shauryaaa.txt', 'r', encoding='utf-8')
f = open('WhatsApp Chat with 2nd year CS -C unofficial group .txt', 'r', encoding='utf-8')
```

```python
data = f.read()
```

```python
print(data)
```

```python
# pattern = '^\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s?(?:am|pm|AM|PM)\s*-\s*(.+?):\s*(.*)$'
# pattern = '^\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s?(?:am|pm|AM|PM)\s-\s(.+?):\s(.*)'
# pattern = '^(\d{1,2}/\d{1,2}/\d{2,4}),\s(\d{1,2}:\d{2})\s?(am|pm|AM|PM)\s*-\s*(.+?):\s*(.*)'
# pattern = '^^(\d{1,2}/\d{1,2}/\d{2,4}),\s\d{1,2}:\d{2}.*?-\s(.*)'
# pattern = '^\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}.*?\b(?:am|pm|AM|PM)\s*-\s*([^:]+:\s.*)$'
# dates pattern = ^(\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s?(?:am|pm|AM|PM))
patternMsg = '\d{1,2}/\d{1,2}/\d{2,4},.*?-\s*'
patternDate = '(\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2})[\u202F\s]?(am|pm|AM|PM)'
```

```python
messages = re.split(patternMsg, data,)[1:]
messages
print(len(messages))
```

```python
dates = re.findall(patternDate, data)
dates
print(len(dates))
```

```python
df = pd.DataFrame({
    "dates": dates,
    "message": messages
})
```

```python
df
```

```python
df["dates"] = df["dates"].astype(str)
```

```python
df['dates'] = df['dates'].str.strip()
df['dates'] = df['dates'].str.replace(r'[()]', '', regex=True)
```

```python
df
```

```python
df['dates'] = df['dates'].str.replace(', pm', ' PM', regex=False)
df['dates'] = df['dates'].str.replace(', am', ' AM', regex=False)
```

```python
df['date'] = df['dates'].str.extract(r"(\d{2}/\d{2}/\d{2}), (\d{1,2}:\d{2}).*?(am|pm)")[0] + ' ' + df['dates'].str.extract(r"(\d{2}/\d{2}/\d{2}), (\d{1,2}:\d{2}).*?(am|pm)")[1] + ' ' + df['dates'].str.extract(r"(\d{2}/\d{2}/\d{2}), (\d{1,2}:\d{2}).*?(am|pm)")[2]
df['date'] = pd.to_datetime(df['date'], format='%d/%m/%y %I:%M %p')
```

```python
df
```

```python
users = []
texts = []

for msg in messages:
    if ':' in msg:
        user, text = msg.split(':', 1)
        users.append(user.strip())
        texts.append(text.strip())
    else:
        users.append("group_notification")
        texts.append(msg.strip())
```

```python
df['user'] = users
```

```python
df['message'] = texts
```

```python
df
```

```python
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month_name()
df['day'] = df['date'].dt.day
df['hour'] = df['date'].dt.hour
df['minute'] = df['date'].dt.minute

df['only_date'] = df['date'].dt.date
df['month_num'] = df['date'].dt.month
df['day_name'] = df['date'].dt.day_name()

```

```python
df.drop(columns=['dates'], inplace=True)
```

```python
df
```

```python
period = []
for hour in df[['day', 'hour']]['hour']:
    if hour == 23:
        period.append(str(hour) + '-' + str(00))
    elif hour == 0:
        period.append(str('00') + '-' + str(hour + 1))
    else: 
        period.append(str(hour) + '-' + str(hour + 1))
```

```python
df['period'] = period
```
